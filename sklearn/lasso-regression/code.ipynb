# Lasso Regression
> Using sklearn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import Lasso, LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
# data generation
from sklearn.datasets import make_regression
x,y = make_regression(n_samples=100, n_features = 1, n_informative = 1, n_targets=1, noise = 20, random_state=13)
```

```python
plt.scatter(x,y,color = 'slateblue', s = 60)
plt.title('True data')
plt.show()
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=2)
```

### Linear Regression

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)

print('intercept: ',lr.intercept_)
print('coef: ',lr.coef_)
print('R2 score: ',r2_score(y_test, y_pred))
```

```python
plt.scatter(x,y,color = 'slateblue', s = 60)
plt.plot(x_test, y_pred, label = 'Linear Regression - BFL', color = 'k')
plt.title('True data')
plt.show()
```

### Lasso Regression

```python
alphas = [0,1,5,10]
plt.figure(figsize=(10,5))
plt.scatter(x,y,color = 'deepskyblue', s = 60, label = 'True data')


for i in alphas:
    las = Lasso(alpha = i)
    las.fit(x_train, y_train)
    y_pred_ls = las.predict(x_test)
    print('intercept: ',las.intercept_)
    print('coef: ',las.coef_)
    print('R2 score: ',r2_score(y_test, y_pred_ls))
    
    plt.plot(x_test, y_pred_ls, label = 'alpha = {}'.format(i), lw = 3)    
plt.legend()
plt.show()
    

```
