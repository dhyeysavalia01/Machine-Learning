## Ridge Regression
> 1 input | 1 output

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.model_selection import train_test_split
````

````python
df = pd.read_csv('study.csv')
df.head()
````

````python
# feature and target
x = df.iloc[:, 0:1]
y = df.iloc[:, -1]
````

````python
# True data scatter plot
plt.scatter(x,y,color = 'deepskyblue', label = 'True data')
plt.xlabel('Attendance Hours')
plt.ylabel('Final Percentage')
plt.legend()
plt.show()
````

## <span style = 'color:orange'>Linear Regression<span>

````python
# simple Linear Regression

# data split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=2)

# Model train
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred_lr = lr.predict(x_test)

print('Intercept: ',lr.intercept_)
print('Coef: ',lr.coef_[0])
print('R2 score: ',r2_score(y_test, y_pred_lr))

# best fit line
plt.scatter(x,y,color = 'deepskyblue', label = 'True data')
plt.plot(x_test, y_pred_lr, color='k', label = 'best fit line')
plt.title('Linear Regression')
plt.xlabel('Attendance Hours')
plt.ylabel('Final Percentage')
plt.legend()
plt.show()
````

## <span style = 'color:orange'>Ridge Regression<span>

````python
r = Ridge(alpha = 0.001)
r.fit(x_train, y_train)
y_pred = r.predict(x_test)

print('Intercept: ',r.intercept_)
print('Coef: ',r.coef_[0])
print('R2 score: ',r2_score(y_test, y_pred))

# best fit line
plt.scatter(x,y,color = 'deepskyblue', label = 'True data')
plt.plot(x_test, y_pred, color='k', label = 'best fit line')
plt.title('Ridge Regression')
plt.xlabel('Attendance Hours')
plt.ylabel('Final Percentage')
plt.legend()
plt.show()
````

````python
ridge_r2 = r2_score(y_test, y_pred)
linear_r2 = r2_score(y_test, y_pred_lr)
print(ridge_r2)
print(linear_r2)
````

````python
ridge_r2 >= linear_r2
````

> Ridge Regression produced a lower ```r2_score``` than Linear Regression because it applies L2 regularization that shrinks coefficients to reduce model complexity, which can slightly decrease the goodness-of-fit on simple datasets with few features.
