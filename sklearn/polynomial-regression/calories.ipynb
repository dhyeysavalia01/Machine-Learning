# <span style='color:orange; font-family:Caveat'>Polynomial Regression<span>
> Multiple Features

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go


from sklearn.metrics import r2_score
from sklearn.model_selection  import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = pd.read_csv('calories.csv')
df.head()
```

```python
# features and target
x = df[['age','exercise_min','heart_rate']]
y = df['calories_burned']
```

```python
# train test split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

## <span style='color:orange; font-family:Caveat'>Linear Regression<span>

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred_lr = lr.predict(x_test)

print('Intercept: ',lr.intercept_)
print('Coef: ',lr.coef_)
print('R2_score: ',r2_score(y_test, y_pred_lr))
```

```python
# meshgrids
x_range = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(), 30)
y_range = np.linspace(df['age'].min(), df['age'].max(), 30)
xx, yy = np.meshgrid(x_range, y_range)

heart_rate_mean = df['heart_rate'].mean()
heart_rate_col = np.full(xx.ravel().shape, heart_rate_mean)

pred_input = np.c_[yy.ravel(), xx.ravel(), heart_rate_col]
zz = lr.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, x='exercise_min', y='age', z='calories_burned', opacity=0.7)
fig.add_surface(x=x_range, y=y_range, z=zz, name='Linear Plane (at Mean HR)', opacity=0.6)
```

<span style='color:navy; font-family:Caveat; font-weight:bold; font-size:100px'>Polynomial Features<span>

```python
poly = PolynomialFeatures(degree = 2, include_bias=True)

x_train_trans = poly.fit_transform(x_train)
x_test_trans = poly.fit_transform(x_test)

lr2 = LinearRegression()
lr2.fit(x_train_trans ,y_train)

y_pred = lr2.predict(x_test_trans)
print('Intercept: ',lr2.intercept_)
print('Coef: ',lr2.coef_)
print('R2_score: ',r2_score(y_test, y_pred))
```

```python
# meshgrids
x_range2 = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(),30)
y_range2 = np.linspace(df['age'].min(), df['age'].max(),30)
xx2, yy2 = np.meshgrid(x_range2, y_range2)

heart_rate_mean2 = df['heart_rate'].mean()
heart_rate_col2 = np.full(xx2.ravel().shape, heart_rate_mean2)

pred_input2 = np.c_[yy2.ravel(), xx2.ravel(), heart_rate_col2]
pred_input2_poly = poly.transform(pred_input2)
zz = lr2.predict(pred_input2_poly).reshape(xx2.shape)


fig2 = px.scatter_3d(df, x = 'exercise_min', y = 'age', z = 'calories_burned')
fig2.add_surface(x = x_range2, y = y_range2, z = zz, name = 'polynomial surface', colorscale = 'plasma')
fig2.show()
```
