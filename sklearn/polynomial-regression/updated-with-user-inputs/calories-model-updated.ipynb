## Polynomial Regression 

```python
import numpy as np
import pandas as pd
import plotly.express as px

from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
```

```python
df = pd.read_csv('calories.csv')
df.head()
```

```python
# features and target
x = df.iloc[:, 0:3]
y = df.iloc[:, -1]
```

```python
# scatter plot of True data
# x = age
# y = exercise_min
# z = calories burned

fig = px.scatter_3d(df, x = df['age'], 
                       y = df['exercise_min'], 
                       z = df['calories_burned'],
                       color = df['calories_burned'],
                       color_continuous_scale='viridis',
                       title = '3d plot')
fig
```

# linear Model training

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=2)
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred_lr = lr.predict(x_test)

print('Intercept: ',lr.intercept_)
print('Coef: ',lr.coef_)
print('R2 Score: ',r2_score(y_test, y_pred_lr))
```

```python
# plane of regression
x_range = np.linspace(df['age'].min(), df['age'].max(), 30)
y_range = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(), 30)

xx, yy = np.meshgrid(x_range, y_range)

heart_mean = df['heart_rate'].mean()
hear_rate_col = np.full(xx.ravel().shape, heart_mean)
pred_input = np.c_[xx.ravel(), yy.ravel(), hear_rate_col]

zz = lr.predict(pred_input).reshape(xx.shape)

fig = px.scatter_3d(df, x = df['age'], 
                       y = df['exercise_min'], 
                       z = df['calories_burned'],
                       color = df['calories_burned'],
                       color_continuous_scale='blues',
                       title = '3d plot')

fig.add_surface(x = x_range, y = y_range, z = zz)

fig.show()
```

# polynomial Features

```python
poly = PolynomialFeatures(degree=2, include_bias=True)
x_train_trans = poly.fit_transform(x_train)
x_test_trans = poly.transform(x_test)

lr2 = LinearRegression()
lr2.fit(x_train_trans, y_train)
y_pred = lr2.predict(x_test_trans)

print('Intercept: ',lr2.intercept_)
print('Coef: ',lr2.coef_)
print('R2 Score: ',r2_score(y_test, y_pred))
```

```python
# plane of regression
x_range = np.linspace(df['age'].min(), df['age'].max(), 30)
y_range = np.linspace(df['exercise_min'].min(), df['exercise_min'].max(), 30)

xx, yy = np.meshgrid(x_range, y_range)

heart_mean = df['heart_rate'].mean()
hear_rate_col = np.full(xx.ravel().shape, heart_mean)
pred_input = np.c_[xx.ravel(), yy.ravel(), hear_rate_col]
pred_input_poly = poly.transform(pred_input)
zz = lr2.predict(pred_input_poly).reshape(xx.shape)

fig = px.scatter_3d(df, x = df['age'], 
                       y = df['exercise_min'], 
                       z = df['calories_burned'],
                       color = df['calories_burned'],
                       color_continuous_scale='blues',
                       title = '3d plot')

fig.add_surface(x = x_range, y = y_range, z = zz)

fig.show()
```
