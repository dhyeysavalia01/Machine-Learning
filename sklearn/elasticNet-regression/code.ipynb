```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Ridge,Lasso,ElasticNet,LinearRegression
```

```python
from sklearn.datasets import make_regression
x,y = make_regression(n_samples = 500, n_features=1, n_informative=1, n_targets=1, noise = 100)
```

```python
plt.scatter(x,y,label = 'True data', s = 60)
```

```python
# split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=2)
```

```python
# linear Regression
lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(x_test)

print('R2 score: ',r2_score(y_test, y_pred))
```

```python
# Ridge Regression (L2)
lr_ridge = Ridge(alpha = 2)
lr_ridge.fit(x_train, y_train)
y_pred_ridge = lr_ridge.predict(x_test)

print('R2 score: ',r2_score(y_test, y_pred_ridge))
```

```python
# Lasso Regression (L1)
lr_lasso = Lasso(alpha = 2)
lr_lasso.fit(x_train, y_train)
y_pred_lasso = lr_lasso.predict(x_test)

print('R2 score: ',r2_score(y_test, y_pred_lasso))
```

```python
# ElasticNet (L1 + L2)
lr_en = ElasticNet(alpha = 2, l1_ratio=0.9)
lr_en.fit(x_train, y_train)
y_pred_en = lr_en.predict(x_test)

print('R2 score: ',r2_score(y_test, y_pred_en))
```
